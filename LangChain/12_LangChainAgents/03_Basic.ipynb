```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.tools import Tool
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
search_tool = DuckDuckGoSearchRun()
result = search_tool.invoke(
    "Top news in India"
)
```

```python
print(result)
```

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
agent = create_agent(
    model = llm,
    tools = [search_tool],
    system_prompt="""
    You are a web research assistant.
    You MUST use the DuckDuckGo search tool to answer every user question.
    Do not answer from your existing knowledge before searching.
    After receiving the search results, provide a clear and helpful answer.
    """
)
```

```python
response = agent.invoke(
    {
        "messages" : [
            ("user", "3 ways to reach goa from delhi")
        ]
    }
)
```

```python
response
```

```python
for message in response["messages"]:
    print("Type:", type(message).__name__)
    print("Content:", message.content)
    print("-" * 50)
```
