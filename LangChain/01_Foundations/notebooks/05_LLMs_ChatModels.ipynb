```python
from dotenv import load_dotenv
import os
load_dotenv()
```

### LLMs

```python
from langchain_openai import OpenAI
from langchain_openai import ChatOpenAI
llm = OpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
prompt = "Explain ML?"
response = llm.invoke(prompt) # Output would be in String
print(response)
```

### ChatModels

```python
from langchain_openai import OpenAI
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
prompt = "Explain ML?"
response = llm.invoke(prompt) 
print(response) # They are behaving differently as compared to the LLMS(Getting extra parameters in the output)
```
