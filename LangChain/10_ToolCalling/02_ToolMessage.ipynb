```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
```

```python
@tool
def multiply(a : int, b : int) -> int:
    """Given 2 numbers a and b. This tool returns their product"""
    return a * b
```

```python
# tool binding
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
llm_with_tools = llm.bind_tools(
    [multiply] # we can give any number of tools (Tool binding)
)
```

```python
query = HumanMessage("Can you multiply 3 with 10")
```

```python
messages = [query]
```

```python
result = llm_with_tools.invoke(messages)
```

```python
result # Ai Message
```

```python
messages.append(result)
```

```python
messages # Human message + AI message
```

```python
tool_result = multiply.invoke(result.tool_calls[0])
```

```python
tool_result # Tool message
```

```python
messages.append(tool_result)
```

```python
messages # Human, AI & Tool Message -> Maintainig a conversation history
```

```python
result = llm_with_tools.invoke(messages) # Now it has a full context of the previous messages so it will give the output as 30
```
