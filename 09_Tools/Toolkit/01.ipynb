```python
from langchain_core.tools import tool
```

```python
@tool
def add(a : int, b : int) -> int:
    """Add two numbers"""
    return a + b
@tool
def multiply(a : int, b : int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit: # Name of your toolkit
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit() # create object for it and then you can use
tools = toolkit.get_tools()
```

```python
for tool in tools:
    print(tool.name, "=>", tool.description)
```
