## Single Message Prompts

### 1 - Static Prompts

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI (
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
system_prompt = """
You are a College Research Assistant.

Your responsibilities are:
- Answer questions about colleges and universities.
- Provide information about admissions, courses, fees,
  placements, scholarships, rankings, and campus life.
- If the question is unrelated to colleges, politely say:
  "I am a College Research Assistant and can only answer
  college-related queries."

Keep your answers concise and informative.
"""
```

```python
user_query = input("Ask your questions:")
```

```python
response = llm.invoke([
    ("system", system_prompt),
    ("human", user_query)
])
```

```python
print(response.content)
```

### 2. Dynamic Prompts

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
llm = ChatOpenAI (
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

### 3. Dynamic Prompt + PromptTemplate

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
```

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
prompt_template = PromptTemplate.from_template(
    "Write 5 points on {topic}." \
    "Use Simple Language and give example also." \
)
prompt = prompt_template.invoke({
    "topic" : "Gaming"
})
response = llm.invoke(prompt)
print(response.content)
```

#### format() Vs invoke()

```python
print(prompt)
print(type(prompt))
prompt2 = prompt_template.format(topic = "Laptop")
print(prompt2)
print(type(prompt2))
```

#### Explicit PromptTemplate

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
prompt_template = PromptTemplate(
    input_variables = ["topic"],
    template = "Write 5 points on {topic}."
)
prompt = prompt_template.invoke({
    "topic" : "Gaming"
})
response = llm.invoke(prompt)
print(response.content)
```

## Multi Message Prompts

### 1. Messages - Manual

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
messages = [
    SystemMessage(
        content="You are a Python tutor. Explain concepts simply and give practical examples."
    ),
    HumanMessage(
        content="Explain decorators in Python."
    )
]

response = llm.invoke(messages)

print(response.content)
```
