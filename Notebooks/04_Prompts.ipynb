### 1 - Static Prompts

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI (
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
system_prompt = """
You are a College Research Assistant.

Your responsibilities are:
- Answer questions about colleges and universities.
- Provide information about admissions, courses, fees,
  placements, scholarships, rankings, and campus life.
- If the question is unrelated to colleges, politely say:
  "I am a College Research Assistant and can only answer
  college-related queries."

Keep your answers concise and informative.
"""
```

```python
user_query = input("Ask your questions:")
```

```python
response = llm.invoke([
    ("system", system_prompt),
    ("human", user_query)
])
```

```python
print(response.content)
```

### 2. Dynamic Prompts

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
llm = ChatOpenAI (
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
```

```python
topic = input("Enter the topic: ")
audience = input("Enter the audience: ")

prompt = f"""
Explain {topic} to a {audience}.
Use simple language and give a practical example.
"""

response = llm.invoke(prompt)

print(response.content)
```
