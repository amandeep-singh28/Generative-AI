```python
from dotenv import load_dotenv
import os
load_dotenv()
```

### Without LangChain

```python
from openai import OpenAI

client = OpenAI(
    api_key=os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1"
)

response = client.responses.create(
    model = "openai/gpt-oss-20b",
    input="Write a short bedtime story about a unicorn."
)
print(response.output_text)
```

### With LangChain - Models Component

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model = 'gemini-2.5-flash',
    api_key = os.getenv("GEMINI_API_KEY")
)
response = llm.invoke("Explain Gen AI in simple words and in short")
print(response.content)
```

* Gemini API keys work only with Gemini models. Groq uses an OpenAI-compatible endpoint, so `ChatOpenAI` can call Groq models when `base_url` points to Groq and the model id exists for your Groq key.

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key=os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1"
)
response = llm.invoke("Explain Gen AI in simple words and in points but in short")
print(response.content)
```
