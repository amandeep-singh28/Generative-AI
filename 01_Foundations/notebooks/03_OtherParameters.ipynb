```python
from dotenv import load_dotenv
import os
load_dotenv()
```

### Temperature - Controls randomness/creativity of the output

```python
from langchain_openai import ChatOpenAI
model_low = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
    temperature = 0
)
model_high = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
    temperature = 1
)
prompt = "Give me only one creative name for a coffee shop(Just only one coffee shop name)."

print("LOW TEMPERATURE:")
print(model_low.invoke(prompt).content)

print("\nHIGH TEMPERATURE:")
print(model_high.invoke(prompt).content)
```

### max_tokens - It puts up a limit to the output

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
    max_tokens = 100
)
response = llm.invoke("Tell me about Virat Kohli in a paragraph?")
print(response.content)
```

### invoke() vs stream()

* Instead of waiting for the complete answer - the application receives pieces as they become available.
* While invoke() - It gives the complete answer

```python
model = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1"
)
for chunk in model.stream("Tell me the concept of Transformer Architecture in Deep Learning."):
    print(chunk.content, end="")
```

### batch() - Combine multiple prompts together

```python
llm = ChatOpenAI(
    model = "openai/gpt-oss-20b",
    api_key = os.getenv("GROK_API_KEY"),
    base_url = "https://api.groq.com/openai/v1",
)
responses = llm.batch([
    "What is AI in short?",
    "What is ML in short?",
    "What is DL in short?"
])
for response in responses:
    print(response.content, end = "\n")
```
